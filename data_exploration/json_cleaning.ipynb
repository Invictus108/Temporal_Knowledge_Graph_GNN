```python
import json

INPUT = "data/wikidata/raw/wikidata_with_description.json"
OUTPUT = "data/wikidata/raw/wikidata_with_description_clean.json"

with open(INPUT, "r", encoding="utf-8") as f:
    data = json.load(f)

cleaned = {}
removed = 0
kept = 0

for year, triples in data.items():
    new_triples = []

    for t in triples:
        head = t.get("head")
        tail = t.get("tail")

        if isinstance(head, dict) and isinstance(tail, dict):
            new_triples.append(t)
            kept += 1
        else:
            removed += 1

    if new_triples:  # only keep non-empty years
        cleaned[year] = new_triples

with open(OUTPUT, "w", encoding="utf-8") as f:
    json.dump(cleaned, f, indent=2, ensure_ascii=False)

print(f"Kept: {kept:,}")
print(f"Removed: {removed:,}")
print(f"Years remaining: {len(cleaned):,}")
```
